#### 정규표현식(정규식)
- 복잡한 문자열을 처리할 때 사용하는 방법
- 파이썬만의 고유한 문법이 아니라 문자열을 처리하는 모든 곳에서 사용하는 일종의 형식 언어

```python
# 주민등록번호 => 뒷자리를 모두 * 문자로 변경

# 파이썬에서 정규식 지원하는 모듈

import re

dir(re)

# re.compile() : 정규식을 작성

pattern = re.compile("ab*")

# 패턴을 특정 문자열이 포함하고 있는가? 
print(pattern.match("abc"))
```

| 메서드 | 목적 |
|-----|----|
|match | 문자열의 처음부터 정규식과 매치되는지 조사|
|search | 문자열 전체를 검색하여 정규식과 매치되는지 조자|
|findall | 정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴|
|finditer | 정규식과 매치되는 모든 문자열(substring)을 이터레이터 객체로 리턴

```python
# [문자] : [] 안에 들어있는 문자들과 매치하느냐? 

p = re.compile("[abc]")

print(p.match("a"))
print(p.match("before"))
print(p.match("dude"))
```

```python
m = p.match("a")

print(f"패턴 시작 : {m.start()}")
print(f"패턴 끝 : {m.end()}")
print(f"패턴과 일치하는 문자열 반환 : {m.group()}")
print(f"일치의 시작,끝 반환 : {m.span()}")

if m:
    print(f"Match found {m.group()}")
else:
    print(f"No Match")
```

- 자주 사용하는 문자 표현식

| 정규표현식 | 설명 |
|-----|----|
|\d|숫자와 매치 [0-9] 와 동일한 표현식|
|\D|숫자가 아닌것과 매치[^0-9]와 동일한 표현식|
|\s|화이트스페이스(스페이스,탭,줄바꿈 등)문자와 매치 [빈칸\t\n\r\f\v] 와 동일한 표현식|
|\S|화이트스페이스(스페이스,탭,줄바꿈 등)문자가 아닌 것과 매치 [^빈칸\t\n\r\f\v] 와 동일한 표현식|
|\w|문자 + 숫자와 매치 [a-zA-Z0-9] 와 동일한 표현식|
|\W|문자 + 숫자가 아닌 것과 매치 [^a-zA-Z0-9] 와 동일한 표현식|

```python
p = re.compile("[0-9]+")
p = re.compile(r"\d+")
print(p.match("012 abc"))


# 원본 문자열에서 0~9가 아닌 문자 찾기
#p = re.compile("[^0-9]+")
p = re.compile(r"\D+")
print(p.match("012 abc"))

# 공백,탭 엔터 문자 찾기
#p = re.compile("[ \t\n\r\f\v]+")
p = re.compile(r"\s+")
print(p.match("012 abc"))

# 공백,탭 엔터 문자가 아닌 문자 찾기
p = re.compile("[^ \t\n\r\f\v]+")
p = re.compile(r"\S+")
print(p.match("012 abc"))


# 대소문자,숫자,_ 포함한 문자 찾기
#p = re.compile("[a-zA-Z0-9_]+")
p = re.compile(r"\w+")
print(p.match("012 abc"))

# 대소문자,숫자,_ 가 아닌 문자 찾기
#p = re.compile("[^a-zA-Z0-9_]+")
p = re.compile(r"\W+")
print(p.match("012 abc"))
```

```python
# re.I : ignorecase

# 영어가 포함되어 잇는지?

p = re.compile("[a-z]+", re.I)
print(p.match("python"))
print(p.match("Python"))
print(p.match("PYTHON"))
```

```python
# re.M : MULTILUNE

data = """python one
life is too short
python two
you need python
python three
"""

# ^ : 시작하는
# python 시작하고 스페이스문자하나영어,숫자,_ 인 문자 찾아줘

p = re.compile(r"^python\s\w+", re.M)
print(p.findall(data))
```

```python
# (?p<이름>패턴) : 그루핑된 문자열에 이름 붙이기
import re
#p = re.compile(r"(\w+)\s+(\d{3}-\d{4}-\d{4})")
p = re.compile(r"(?P<name>\w+)\s+(?P<tell>\d{3}-\d{4}-\d{4})")
m = p.search("park 010-1234-5678")
print(m)
print(m.group())
print(m.group("name"))
print(m.group("tell"))
```

```python
import re

error = 0
info = 0
err_logs = []
daily_error = {}
pattern = re.compile(r"(?P<date>\d{4}-\d{2}-\d{2})\s(?P<level>INFO|ERROR)\s(?P<message>.+)")





for log in logs:
    m = pattern.search(log)

    if not m:
        continue

    date = m.group("date")
    level = m.group("level")
    message = m.group("message")

    if level == "INFO":
        info += 1

    elif level == "ERROR":
        error += 1
        err_logs.append(message)
        daily_error[date] = daily_error.get(date, 0) + 1

print("ERROR 목록")
print("-" * 20)

for message in err_logs:
    print(message)

print("-" * 20)
print(f"INFO : {info}")
print(f"ERROR : {error}")

for key, value in daily_error.items():
    print(f"{key} : {value}")
```
